# <font color='blue'> Biological Inspiration and the Artificial Neuron </font>

The development of artificial neural networks was originally motivated by the remarkable computational capabilities of the human brain. Although the brain consists of billions of biological neurons connected through trillions of synapses, it is capable of performing complex tasks such as perception, learning, memory, reasoning, and decision-making while consuming remarkably little energy. Early researchers sought to understand whether some of these computational principles could be reproduced using mathematical models.

Artificial neural networks are therefore inspired by the organization of biological nervous systems rather than attempting to replicate them exactly. The individual computational units used in modern neural networks, called **artificial neurons**, are highly simplified mathematical abstractions that capture only the essential idea of combining multiple inputs to produce an output.

This chapter introduces the biological neuron, explains how its structure inspired artificial neural networks, presents the McCulloch–Pitts neuron model, and develops the mathematical formulation of the artificial neuron that forms the foundation of modern deep learning.

The human brain contains approximately

```
86 billion neurons
```

connected by an estimated

```
100–500 trillion synapses.
```

Despite consuming only about

```
20 watts
```

* visual perception,

* learning,

These capabilities inspired the earliest neural network researchers.

## <font color='orange'> The Biological Neuron </font>

A biological neuron consists of several main components.

```
          Dendrites

        \   |   /

         \  |  /

      [ Cell Body ]

             |

          Axon

             |

        Synaptic Terminals
```

Each component has a distinct function.

| Component | Function |
| :--- | :--- |
| Dendrites | Receive signals from other neurons |
| Cell Body (Soma) | Integrates incoming signals |
| Axon | Transmits electrical impulses |
| Synapses | Connect one neuron to another |

Together, billions of neurons communicate through electrical and chemical signals to form highly complex neural circuits.

---



## <font color='orange'> How Biological Neurons Communicate </font>

The communication process can be summarized as

```
Incoming Signals

↓

Dendrites

↓

Cell Body

↓

Action Potential?

↓

Axon

↓

Synapse

↓

Next Neuron
```

If the combined input exceeds a certain threshold, the neuron generates an electrical impulse, known as an **action potential**, which travels along the axon and influences other neurons.

---



## <font color='orange'> Inspiration for Artificial Intelligence </font>

Early researchers asked a simple but profound question:

> **Can learning emerge from large networks of simple computational units?**

This question led to the development of mathematical neuron models that abstract away the biological complexity while retaining the essential idea of combining multiple inputs to produce an output. Modern deep learning is the result of decades of refinement of these mathematical abstractions.

---



## <font color='orange'> The McCulloch–Pitts Neuron </font>

In 1943, Warren McCulloch and Walter Pitts proposed one of the earliest mathematical models of a neuron.
The model operates in three steps.

```
Inputs

↓

Weighted Sum

↓

Threshold

↓

Output
```

The neuron produces either

```
0

or

1
```

depending on whether the weighted input exceeds a specified threshold. Although extremely simple, this model demonstrated that logical computations could be performed using interconnected artificial neurons.

---



## <font color='orange'> From Biology to Mathematics </font>

The correspondence between biological and artificial neurons is

| Biological System | Artificial Model |
| :--- | :--- |
| Dendrites | Input features |
| Synapses | Weights |
| Cell body | Weighted summation |
| Action potential | Activation function |
| Axon | Output |

This analogy provides intuition, although modern artificial neurons are vastly simpler than real biological neurons.

# <font color='blue'> The Artificial Neuron </font>

An artificial neuron receives multiple numerical inputs.

```
x₁

x₂

x₃

⋮

xₙ
```

Each input has an associated weight

```
w₁

w₂

w₃

⋮

wₙ
```

The neuron combines these inputs to produce a single output.

---



## <font color='orange'> Weighted Sum </font>

The first computation performed by the neuron is a weighted sum. Mathematically,

$$
\boxed{
z
=
\sum_{i=1}^{n}
w_i x_i
+
b
}
$$

where

* $x_i$ are the input features,
* $w_i$ are the weights,
* $b$ is the bias term.

This equation forms the foundation of every modern neural network.

---



### <font color='violet'> Why Are Weights Needed? </font>

Not every input contributes equally. Weights determine how strongly each input influences the output.
Conceptually,

<font color='blue'> Case I </font>
```
Large Positive Weight

↓

Strong Positive Influence
```
<font color='blue'> Case II </font>
```
Negative Weight

↓

Suppresses Input
```
<font color='blue'> Case III </font>
```
Weight Near Zero

↓

Little Influence
```

During training, the network automatically learns the optimal values of these weights.

---



### <font color='violet'> The Role of the Bias </font>

The bias acts as an adjustable offset. Without a bias, the weighted sum would always pass through the origin. Including the bias allows the neuron greater flexibility, making it easier to represent a wider variety of functions. The bias is analogous to the intercept term in linear regression.

---



## <font color='orange'> Activation </font>

After computing the weighted sum, the neuron applies an **activation function**. Conceptually,

```
Inputs

↓

Weighted Sum

↓

Activation Function

↓

Output
```

The activation function determines how strongly the neuron responds to its inputs. Modern neural networks employ activation functions such as

* Sigmoid,
* Tanh,
* ReLU,

which will be discussed in detail later.

---



# <font color='blue'> A Network of Artificial Neurons </font>

Individual neurons are useful, but their true power emerges when many neurons are connected together. Conceptually,

```
Input Layer

↓

Hidden Layer

↓

Hidden Layer

↓

Output Layer
```

Each neuron performs a simple computation, yet together they can approximate remarkably complex functions. This layered organization gives rise to **deep neural networks**.

---



## <font color='orange'> Limitations of the Biological Analogy </font>

Although inspired by biology, modern artificial neural networks differ substantially from real brains.
For example,

* biological neurons communicate through complex electrochemical processes,
* real neurons exhibit highly dynamic behaviour,
* the brain contains many specialised cell types beyond neurons,
* learning in biological systems is not equivalent to backpropagation.

Artificial neural networks should therefore be viewed as useful mathematical abstractions rather than faithful biological models.

| Concept | Description |
| :--- | :--- |
| Biological Neuron | Fundamental computational unit of the nervous system |
| Dendrites | Receive signals from other neurons |
| Axon | Transmits signals to other neurons |
| Synapses | Connections between neurons |
| McCulloch–Pitts Neuron | Early mathematical neuron model based on threshold logic |
| Artificial Neuron | Simplified computational model combining weighted inputs |
| Weighted Sum | Linear combination of inputs and weights plus a bias |
| Activation Function | Transforms the weighted sum into the neuron's output |
| Main Insight | Complex behaviour emerges from networks of many simple computational units |

> **Key Insight:** Artificial neural networks were inspired by the structure of biological nervous systems but are fundamentally mathematical models rather than biological simulations. An artificial neuron receives multiple inputs, computes a weighted sum, adds a bias, and applies an activation function to produce an output. Although individually simple, large networks of interconnected artificial neurons can learn highly complex nonlinear relationships, forming the foundation of modern deep learning.
